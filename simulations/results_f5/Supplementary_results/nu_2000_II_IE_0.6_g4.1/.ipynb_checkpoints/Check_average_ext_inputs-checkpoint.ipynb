```python
from brian2 import *
prefs.codegen.target = 'numpy'  # use the Python fallback (can also use cython)
import numpy as np
import matplotlib.pyplot as plt
import scipy.io as sio
#from brian2tools import *
np.random.seed(0) # Set the random seed
```

```python
 
index_list=1
list_index=1
I_rho = 0.6
E_rho = 0.6

# Grid and neuron parameters (unchanged)
Size_x=2050
Size_y=Size_x
Volume=Size_x*Size_y*umeter**2

grid_dist_r=400*umeter
grid_nr=int((Size_x-400 + grid_dist_r/umeter)//(grid_dist_r/umeter))
rows_r, cols_r = grid_nr,grid_nr

Total_nr_neurons=28000*(Volume/mm**2)
N_E=int(Total_nr_neurons*0.8)   # Pyramidal cells
N_I=int(Total_nr_neurons*0.2)   # Interneurons
N_R=grid_nr**2                  # Recording sites
N_Pr=1500                       # Nr of neurons whose spike trains I will record

# Setting up recording sites (unchanged)
eqs_r='''x : meter
         y : meter
         index : integer (constant)
'''
Rec = NeuronGroup(rows_r * cols_r, eqs_r)
Rec.x = '(i // rows_r) * grid_dist_r - (rows_r-1)/2.0  * grid_dist_r'
Rec.y = '(i % cols_r) * grid_dist_r - (cols_r-1)/2.0  * grid_dist_r'

sio.savemat('Rec_x.mat', {'data': Rec.x[:]})
sio.savemat('Rec_y.mat', {'data': Rec.y[:]})

##################
# External input #
##################

nu_preparation=1700*Hz
nu_execution=5000-nu_preparation/Hz
tau_x=50*ms
duration_ext=2500*ms
sigma_n = 5 * Hz/sqrt(ms)  # or numerically: 250 Hz/sqrt(s)
tau_n=150*ms

# Waveform parameters
nu_baseline = 1750 * Hz      # baseline rate
A_gauss = 500 * Hz           # Gaussian bump amplitude
t0 = 1200 * ms               # sigmoid midpoint time
t0g = 1500 * ms              # Gaussian center time
sigma = 200 * ms             # Gaussian width
A_sigm = 6000 * Hz - nu_baseline   # sigmoid step height
k = 0.08 / ms                # sigmoid slope (steepness)

# Sigmoid parameters (new)
beta = 0.1  # Temperature parameter (adjust as needed)
t_mid = 1200*ms  # Midpoint of sigmoid (time when sigmoid is at 0.5)
sigmoid_scale = nu_execution  # Scale factor for sigmoid amplitude

##############
# Duration   #
##############

number_iterations=1
duration=2500*ms
duration_ext=number_iterations*duration

##################
# Single neuron  #
##################

V_th=18*mV
v_reset=11*mV

tau_mE=(20)*ms
tau_mI=(10)*ms
tau_ref_E=2*ms
tau_ref_I=1*ms
tau_rAI=0.7*ms
tau_rAE=0.7*ms
tau_dAE =3.5*ms
tau_dAI =3.5*ms

tau_rG=1*ms
tau_dG=20*ms

#############################################################################
#  Synapses
#############################################################################

lam_dist=[1.7*550,1.7*350,1.7*350,1.7*350]
EEprob_z=0.15

J_ext_I=0.98
J_ext_E=0.55

g=4.2
J_EE=0.42/4
J_EI=0.7/4
J_IE=g*J_EE
J_II=g*J_EI

##################
start_scope()

# First, let's pre-generate the global rate time series
# instead of using a NeuronGroup
dt_val = defaultclock.dt
steps = int(duration_ext/dt_val)
time_array = np.arange(steps) * dt_val
x_global = np.zeros(steps)
en_global = np.zeros(steps)
nu_global = np.zeros(steps)

# Sigmoid function
def sigmoid(t, beta, t_mid, scale):
    """
    Calculates sigmoid function with temperature parameter beta
    t: time
    beta: temperature parameter (controls steepness)
    t_mid: time at midpoint of sigmoid
    scale: maximum amplitude of sigmoid
    """
    return scale / (1 + np.exp(-beta * (t - t_mid)/ms))

# Generate the time series with simple Euler integration
# Using an actual sigmoid function with temperature parameter
# BEFORE the loop
 
# Pre-allocate pure-float arrays
x_global  = np.zeros(steps) * Hz
nu_global = np.zeros(steps) * Hz
en_global = np.zeros(steps) * Hz

en_global[0] = np.random.randn() * sigma_n * np.sqrt(tau_n)

for i in range(steps):
    t = time_array[i]
    # 1) Build the waveform as a Quantity in Hz
    gaussian = A_gauss * np.exp(-((t - t0g)**2) / (2*sigma**2))
    sigmoid  = A_sigm  / (1 + np.exp(-k * (t - t0)))
    rate_q   = nu_baseline + gaussian + sigmoid      # Quantity (Hz)
 
    # 3) Update your unitless noise
    if i > 0:
        en_global[i] = en_global[i-1] + (-en_global[i-1]/tau_n) * dt_val + sigma_n * np.sqrt(2 * dt_val) * np.random.randn()
    nu_global[i] =  rate_q + en_global[i]
 
```
