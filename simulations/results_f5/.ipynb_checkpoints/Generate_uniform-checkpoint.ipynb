```python

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# SETTINGS
# ============================================================

DATA_DIRS = [
    Path("results_f5_g3.6_rho0.6_nu0.91/wave_direction_analysis"),
    # Path("results_f5_g3.6_rho0.6_nu1.0/wave_direction_analysis"),
]

LABELS = [
    r"$\nu_{\mathrm{ext}}/\nu_\theta = 0.91$",
    # r"$\nu_{\mathrm{ext}}/\nu_\theta = 1.0$",
]

COLORS = ["#163B65", "#C44E52"]

THRESHOLD_METHOD = "fixed"  # "fixed" or "statistical"
HISTOGRAM_NORMALIZATION = "probability"  # "probability" or "count"

OUTPUT_DIR = Path("wave_direction_comparison")

# Use the same radial scale for observed and uniform
# WITHIN each condition, but different scales BETWEEN conditions.
MATCH_UNIFORM_SCALE = True

RANDOM_SEED = 42

# Keep text editable in Inkscape
plt.rcParams["svg.fonttype"] = "none"

rng = np.random.default_rng(RANDOM_SEED)

if len(DATA_DIRS) != len(LABELS) or len(DATA_DIRS) > len(COLORS):
    raise ValueError("DATA_DIRS, LABELS and COLORS must match")

if HISTOGRAM_NORMALIZATION not in ("probability", "count"):
    raise ValueError("Invalid normalization")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# LOAD SAVED HISTOGRAM
# ============================================================

def load_histogram(folder):

    file = folder / f"wave_direction_{THRESHOLD_METHOD}.npz"

    if not file.is_file():
        raise FileNotFoundError(file)

    with np.load(file) as data:
        counts = data["histogram_counts"].astype(float)
        edges = data["histogram_edges_rad"].astype(float)
        directions = data["direction_rad"].astype(float)

    print(f"Loaded: {file}")
    print(f"Events: {int(counts.sum())}")

    return counts, edges, directions


# ============================================================
# HISTOGRAM FUNCTIONS
# ============================================================

def calculate_heights(counts):

    heights = counts.astype(float).copy()

    if HISTOGRAM_NORMALIZATION == "probability":
        if heights.sum() > 0:
            heights /= heights.sum()

    return heights


def histogram_from_angles(angles, edges):

    angles = np.asarray(angles, dtype=float)
    angles = angles[np.isfinite(angles)]

    # Match the angular interval of the original saved histogram
    wrapped = (angles - edges[0]) % (2*np.pi) + edges[0]

    counts, _ = np.histogram(wrapped, bins=edges)

    return counts


# ============================================================
# POLAR PLOT
# ============================================================

def plot_polar_histogram(
    heights, edges, title, color, filename, radial_max=None
):

    centers = (edges[:-1] + edges[1:]) / 2
    widths = np.diff(edges)

    fig, ax = plt.subplots(
        figsize=(6, 6),
        subplot_kw={"projection": "polar"}
    )

    ax.bar(
        centers,
        heights,
        width=widths,
        color=color,
        edgecolor="black",
        linewidth=0.7,
        alpha=0.85
    )

    ax.set_theta_zero_location("E")
    ax.set_theta_direction(1)

    ax.set_thetagrids(
        [0, 90, 180, 270],
        ["0° (+x)", "90° (+y)", "180° (−x)", "270° (−y)"]
    )

    ax.set_rlabel_position(45)

    if radial_max is not None:
        ax.set_ylim(0, radial_max)

    ax.set_title(title, pad=25)

    fig.tight_layout()

    fig.savefig(
        OUTPUT_DIR / filename,
        format="svg",
        bbox_inches="tight"
    )

    return fig


# ============================================================
# LOAD ALL DATASETS
# ============================================================

histograms = [load_histogram(folder) for folder in DATA_DIRS]

# Common angular bins for all conditions
edges = histograms[0][1]


# ============================================================
# PLOT OBSERVED AND UNIFORM DISTRIBUTIONS
# ============================================================

for i, (saved_counts, saved_edges, directions) in enumerate(histograms):

    # Use saved counts if bins match; otherwise rebin directions
    if np.allclose(saved_edges, edges) and len(saved_edges) == len(edges):
        counts = saved_counts
    else:
        counts = histogram_from_angles(directions, edges)

    n_events = int(counts.sum())

    # --------------------------------------------
    # Observed distribution
    # --------------------------------------------

    heights = calculate_heights(counts)

    # --------------------------------------------
    # Uniform reference: same number of events
    # --------------------------------------------

    uniform_angles = rng.uniform(
        0,
        2*np.pi,
        size=n_events
    )

    uniform_counts = histogram_from_angles(
        uniform_angles,
        edges
    )

    uniform_heights = calculate_heights(uniform_counts)

    # --------------------------------------------
    # Radial limits
    # --------------------------------------------

    observed_max = max(heights.max(), 1e-6) * 1.15
    uniform_max = max(uniform_heights.max(), 1e-6) * 1.15

    if MATCH_UNIFORM_SCALE:
        common_max = max(observed_max, uniform_max)
        observed_max = common_max
        uniform_max = common_max

    # --------------------------------------------
    # Plot observed
    # --------------------------------------------

    plot_polar_histogram(
        heights,
        edges,
        f"{LABELS[i]}\n{n_events} wave events",
        COLORS[i],
        f"direction_{THRESHOLD_METHOD}_{i+1}.svg",
        radial_max=observed_max
    )

    # --------------------------------------------
    # Plot uniform
    # --------------------------------------------

    plot_polar_histogram(
        uniform_heights,
        edges,
        f"Uniform reference\n{n_events} samples",
        "#888888",
        f"uniform_reference_{i+1}.svg",
        radial_max=uniform_max
    )

    print(
        f"Condition {i+1}: "
        f"{n_events} observed events, "
        f"{n_events} uniform samples"
    )

plt.show()

print(f"\nSaved SVG figures to {OUTPUT_DIR.resolve()}")
```
