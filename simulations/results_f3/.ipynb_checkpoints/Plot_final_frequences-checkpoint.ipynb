```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat, savemat
```

```python
# MEAN FIELD

myname = "26caStability_g_fwdtest_3.625_static.mat"

lc = loadmat(
    myname,
    squeeze_me=True,
    struct_as_record=False
)

list_varying_param = np.asarray(lc["list_varying_param"]).squeeze()
local_minima_all = np.atleast_1d(lc["local_minima_all"])

all_param_list = []
all_minima_list = []

for i, min_h in enumerate(local_minima_all):

    # Convert MATLAB cell content to numeric array
    min_h = np.asarray(min_h, dtype=float)
    min_h = np.atleast_2d(min_h)

    nr_minima = min_h.shape[0]

    params_h = np.full(
        nr_minima,
        list_varying_param[i]
    )

    all_param_list.append(params_h)
    all_minima_list.append(min_h)

all_param = np.concatenate(all_param_list)
all_minima = np.vstack(all_minima_list)

print("all_param shape:", all_param.shape)
print("all_minima shape:", all_minima.shape)
```

```python
# Excitatory population
ld_E = loadmat(
    "Oscillation_combined_rates_E.mat",
    squeeze_me=True
)

ld_E_L = loadmat(
    "Oscillation_combined_rates_E_large.mat",
    squeeze_me=True
)

# Inhibitory population
ld_I = loadmat(
    "Oscillation_combined_rates_I.mat",
    squeeze_me=True
)

ld_I_L = loadmat(
    "Oscillation_combined_rates_I_large.mat",
    squeeze_me=True
)

input_drive = np.asarray(ld_E["input_drive"], dtype=float).squeeze()


# --------------------------------
# E population
# --------------------------------

all_freq_E = np.vstack([
    np.atleast_2d(np.asarray(ld_E["all_mean_freq"], dtype=float)),
    np.atleast_2d(np.asarray(ld_E_L["all_mean_freq"], dtype=float)),
])

mean_freq_E = np.mean(all_freq_E, axis=0)
std_freq_E = np.std(all_freq_E, axis=0, ddof=1)


# --------------------------------
# I population
# --------------------------------

all_freq_I = np.vstack([
    np.atleast_2d(np.asarray(ld_I["all_mean_freq"], dtype=float)),
    np.atleast_2d(np.asarray(ld_I_L["all_mean_freq"], dtype=float)),
])

mean_freq_I = np.mean(all_freq_I, axis=0)
std_freq_I = np.std(all_freq_I, axis=0, ddof=1)

print("E:", all_freq_E.shape)
print("I:", all_freq_I.shape)
```

```python
fig, ax = plt.subplots(figsize=(6, 5))

# Mean-field prediction
ax.plot(
    all_param,
    all_minima[:, 1],
    "k",
    linewidth=2,
    label="Stability prediction"
)

# E population
ax.errorbar(
    input_drive,
    mean_freq_E,
    yerr=std_freq_E,
    fmt="o",
    capsize=3,
    label="E rate"
)

# I population
ax.errorbar(
    input_drive,
    mean_freq_I,
    yerr=std_freq_I,
    fmt="s",
    capsize=3,
    label="I rate"
)

ax.set_xlabel(r"$\nu_{\mathrm{ext}} / \nu_{\mathrm{th}}$")
ax.set_ylabel("Dominant frequency [Hz]")
ax.legend()

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 5))

ax.errorbar(
    input_drive,
    mean_freq_large,
    yerr=std_freq_large,
    fmt=".k",
    markersize=20
)

ax.plot(
    all_param,
    all_minima[:, 1],
    "k",
    linewidth=2
)

ax.set_xlabel(r"$\nu_{\mathrm{ext}} / \nu_{\mathrm{th}}$")
ax.set_ylabel("Dominant frequency [Hz]")

plt.tight_layout()
plt.show()
```

```python
savemat(
    "Oscillation_final_rates_E.mat",
    {
        "input_drive": input_drive,
        "mean_freq": mean_freq,
        "std_freq": std_freq,
    }
)
```
