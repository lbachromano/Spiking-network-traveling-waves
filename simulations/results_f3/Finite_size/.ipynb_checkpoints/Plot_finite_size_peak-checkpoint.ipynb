```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat


# ============================================================
# Load
# ============================================================

data_085 = loadmat("FiniteSize_nu0.85_g3.18.mat")
data_160 = loadmat("FiniteSize_nu1.6_g3.18.mat")

N_085 = np.asarray(
    data_085["N_total"]
).squeeze()

N_160 = np.asarray(
    data_160["N_total"]
).squeeze()


ratio_085 = np.asarray(
    data_085["peak_background_ratio"]
)

ratio_160 = np.asarray(
    data_160["peak_background_ratio"]
)


# Make sure rows correspond to N
if ratio_085.shape[0] != len(N_085):
    ratio_085 = ratio_085.T

if ratio_160.shape[0] != len(N_160):
    ratio_160 = ratio_160.T


# ============================================================
# Mean and SD across independent realizations
# ============================================================

mean_ratio_085 = np.nanmean(ratio_085, axis=1)
std_ratio_085 = np.nanstd(
    ratio_085,
    axis=1,
    ddof=1,
)

mean_ratio_160 = np.nanmean(ratio_160, axis=1)
std_ratio_160 = np.nanstd(
    ratio_160,
    axis=1,
    ddof=1,
)


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(6, 4.5))

plt.errorbar(
    N_085, mean_ratio_085, yerr=std_ratio_085,
    fmt="o-", capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 0.85$"
)

plt.errorbar(
    N_160, mean_ratio_160, yerr=std_ratio_160,
    fmt="o-", capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 1.6$"
)

plt.yscale("log")
plt.xlabel("Number of neurons, N")
plt.ylabel("Dominant peak / spectral background")
plt.legend()
plt.grid(True, which="both")
plt.tight_layout()
plt.show()
```

```python
freq_085 = np.asarray(
    data_085["peak_freq"]
)

freq_160 = np.asarray(
    data_160["peak_freq"]
)

if freq_085.shape[0] != len(N_085):
    freq_085 = freq_085.T

if freq_160.shape[0] != len(N_160):
    freq_160 = freq_160.T


mean_freq_085 = np.nanmean(freq_085, axis=1)
std_freq_085 = np.nanstd(freq_085, axis=1, ddof=1)

mean_freq_160 = np.nanmean(freq_160, axis=1)
std_freq_160 = np.nanstd(freq_160, axis=1, ddof=1)


plt.figure(figsize=(6, 4.5))

plt.errorbar(
    N_085,
    mean_freq_085,
    yerr=std_freq_085,
    fmt="o-",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 0.85$",
)

plt.errorbar(
    N_160,
    mean_freq_160,
    yerr=std_freq_160,
    fmt="o-",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 1.6$",
)

plt.xlabel("Number of neurons, N")
plt.ylabel("Dominant peak frequency (Hz)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat


# ============================================================
# LOAD DATA
# ============================================================

data_085 = loadmat("FiniteSize_nu0.85_g3.18.mat")
data_160 = loadmat("FiniteSize_nu1.6_g3.18.mat")

N_085 = np.asarray(data_085["N_total"]).squeeze()
N_160 = np.asarray(data_160["N_total"]).squeeze()

ratio_085 = np.asarray(
    data_085["peak_background_ratio"],
    dtype=float,
)

ratio_160 = np.asarray(
    data_160["peak_background_ratio"],
    dtype=float,
)


# Make sure:
# rows    = network sizes
# columns = independent realizations
if ratio_085.shape[0] != len(N_085):
    ratio_085 = ratio_085.T

if ratio_160.shape[0] != len(N_160):
    ratio_160 = ratio_160.T


# ============================================================
# MEAN AND SD ACROSS REALIZATIONS
# ============================================================

mean_ratio_085 = np.nanmean(ratio_085, axis=1)
std_ratio_085 = np.nanstd(
    ratio_085,
    axis=1,
    ddof=1,
)

mean_ratio_160 = np.nanmean(ratio_160, axis=1)
std_ratio_160 = np.nanstd(
    ratio_160,
    axis=1,
    ddof=1,
)


# ============================================================
# SANITY CHECK
# ============================================================

print("Peak/background ratio")
print()
print("nu_ext / nu_theta = 0.85")
print(f"  minimum individual value = {np.nanmin(ratio_085):.4g}")
print(f"  maximum individual value = {np.nanmax(ratio_085):.4g}")
print(f"  means across N            = {mean_ratio_085}")

print()

print("nu_ext / nu_theta = 1.6")
print(f"  minimum individual value = {np.nanmin(ratio_160):.4g}")
print(f"  maximum individual value = {np.nanmax(ratio_160):.4g}")
print(f"  means across N            = {mean_ratio_160}")
```
