```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat

# ------------------------------------------------------------
# Load results
# ------------------------------------------------------------

data_085 = loadmat("FiniteSize_nu0.85_g3.18.mat")
data_160 = loadmat("FiniteSize_nu1.6_g3.18.mat")

N_085 = np.asarray(data_085["N_total"]).squeeze()
mean_085 = np.asarray(data_085["mean_osc_strength"]).squeeze()
std_085 = np.asarray(data_085["std_osc_strength"]).squeeze()

N_160 = np.asarray(data_160["N_total"]).squeeze()
mean_160 = np.asarray(data_160["mean_osc_strength"]).squeeze()
std_160 = np.asarray(data_160["std_osc_strength"]).squeeze()


# ------------------------------------------------------------
# Transform x-axis for finite-size scaling
# ------------------------------------------------------------

x_085 = 1.0 / np.sqrt(N_085)
x_160 = 1.0 / np.sqrt(N_160)

# Linear fits:
# A_osc = a / sqrt(N) + b

p_085 = np.polyfit(x_085, mean_085, 1)
p_160 = np.polyfit(x_160, mean_160, 1)

x_fit = np.linspace(
    min(x_085.min(), x_160.min()),
    max(x_085.max(), x_160.max()),
    200,
)

y_fit_085 = np.polyval(p_085, x_fit)
y_fit_160 = np.polyval(p_160, x_fit)


# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))


# ============================================================
# Panel 1: Oscillation strength vs N
# ============================================================

axes[0].errorbar(
    N_085,
    mean_085,
    yerr=std_085,
    fmt="o-",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 0.85$",
)

axes[0].errorbar(
    N_160,
    mean_160,
    yerr=std_160,
    fmt="o-",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 1.6$",
)

axes[0].set_xlabel("Number of neurons, N")
axes[0].set_ylabel("Oscillation strength (Hz)")
axes[0].grid(True)
axes[0].legend()


# ============================================================
# Panel 2: Oscillation strength vs 1/sqrt(N)
# ============================================================

axes[1].errorbar(
    x_085,
    mean_085,
    yerr=std_085,
    fmt="o",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 0.85$",
)

axes[1].errorbar(
    x_160,
    mean_160,
    yerr=std_160,
    fmt="o",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 1.6$",
)

axes[1].plot(
    x_fit,
    y_fit_085,
    "-",
    label=(
        rf"0.85 fit: "
        rf"$A={p_085[0]:.3g}/\sqrt{{N}}"
        rf"{p_085[1]:+.3g}$"
    ),
)

axes[1].plot(
    x_fit,
    y_fit_160,
    "-",
    label=(
        rf"1.6 fit: "
        rf"$A={p_160[0]:.3g}/\sqrt{{N}}"
        rf"{p_160[1]:+.3g}$"
    ),
)

axes[1].set_xlabel(r"$1/\sqrt{N}$")
axes[1].set_ylabel("Oscillation strength (Hz)")
axes[1].grid(True)
axes[1].legend()

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Print fit parameters
# ------------------------------------------------------------

print("nu_ext / nu_theta = 0.85")
print(f"  slope a     = {p_085[0]:.6g}")
print(f"  intercept b = {p_085[1]:.6g}")

print()

print("nu_ext / nu_theta = 1.6")
print(f"  slope a     = {p_160[0]:.6g}")
print(f"  intercept b = {p_160[1]:.6g}")
```
