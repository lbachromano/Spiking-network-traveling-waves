```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat


# ============================================================
# SETTINGS
# ============================================================

N_BOOT = 10000
RANDOM_SEED = 12345


# ============================================================
# LOAD DATA
# ============================================================

def load_finite_size(filename):
    data = loadmat(filename)

    N = np.asarray(data["N_total"], dtype=float).squeeze()
    A = np.asarray(data["LFP_osc_strength"], dtype=float)
    peak_freq = np.asarray(data["LFP_peak_freq"], dtype=float)

    # We expect:
    # rows    = network sizes
    # columns = independent realizations

    if A.shape[0] != len(N) and A.shape[1] == len(N):
        A = A.T

    if peak_freq.shape[0] != len(N) and peak_freq.shape[1] == len(N):
        peak_freq = peak_freq.T

    if A.shape[0] != len(N):
        raise ValueError(
            f"Inconsistent dimensions: N has length {len(N)}, "
            f"but osc_strength has shape {A.shape}"
        )

    return N, A, peak_freq


N_085, A_085, freq_085 = load_finite_size("FiniteSize_nu0.85_g3.18.mat")
N_160, A_160, freq_160 = load_finite_size("FiniteSize_nu1.6_g3.18.mat")

print("0.85:", A_085.shape)
print("1.6 :", A_160.shape)
```

```python
# ============================================================
# FITTING FUNCTIONS
# ============================================================

def fit_free_intercept(N, A):
    """
    Fit:
        A = a / sqrt(N) + b

    Uses all independent realizations.
    """

    x = 1.0 / np.sqrt(N)

    # Repeat each x once for each realization
    X = np.repeat(x, A.shape[1])
    Y = A.reshape(-1)

    valid = np.isfinite(X) & np.isfinite(Y)
    X = X[valid]
    Y = Y[valid]

    # y = a*x + b
    a, b = np.polyfit(X, Y, 1)

    prediction = a * X + b
    rss = np.sum((Y - prediction) ** 2)

    return a, b, rss


def fit_zero_intercept(N, A):
    """
    Fit constrained finite-size model:
        A = a / sqrt(N)

    i.e. b = 0.
    """

    x = 1.0 / np.sqrt(N)

    X = np.repeat(x, A.shape[1])
    Y = A.reshape(-1)

    valid = np.isfinite(X) & np.isfinite(Y)
    X = X[valid]
    Y = Y[valid]

    # Least-squares fit through origin
    a = np.sum(X * Y) / np.sum(X ** 2)

    prediction = a * X
    rss = np.sum((Y - prediction) ** 2)

    return a, rss


def aic_from_rss(rss, n, k):
    """
    Gaussian AIC, ignoring constants common to both models.
    """
    return n * np.log(rss / n) + 2 * k
```

```python
# ============================================================
# FIT BOTH MODELS
# ============================================================

def analyse_condition(N, A, label):

    # Model 1:
    # A = a/sqrt(N)
    a0, rss0 = fit_zero_intercept(N, A)

    # Model 2:
    # A = a/sqrt(N) + b
    a, b, rss1 = fit_free_intercept(N, A)

    n_obs = np.sum(np.isfinite(A))

    aic0 = aic_from_rss(rss0, n_obs, k=1)
    aic1 = aic_from_rss(rss1, n_obs, k=2)

    print()
    print("=" * 60)
    print(label)
    print("=" * 60)

    print("\nConstrained finite-size model:")
    print("A = a / sqrt(N)")
    print(f"a   = {a0:.6g}")
    print(f"RSS = {rss0:.6g}")
    print(f"AIC = {aic0:.3f}")

    print("\nFree-intercept model:")
    print("A = a / sqrt(N) + b")
    print(f"a   = {a:.6g}")
    print(f"b   = {b:.6g}")
    print(f"RSS = {rss1:.6g}")
    print(f"AIC = {aic1:.3f}")

    print(f"\nDelta AIC = AIC(b=0) - AIC(free) = {aic0 - aic1:.3f}")

    return {
        "a0": a0,
        "rss0": rss0,
        "a": a,
        "b": b,
        "rss1": rss1,
        "aic0": aic0,
        "aic1": aic1,
    }


result_085 = analyse_condition(
    N_085,
    A_085,
    r"$\nu_{ext}/\nu_\theta = 0.85$"
)

result_160 = analyse_condition(
    N_160,
    A_160,
    r"$\nu_{ext}/\nu_\theta = 1.6$"
)
```

```python
# ============================================================
# BOOTSTRAP THE INTERCEPT b
# ============================================================

def bootstrap_intercept(
    N,
    A,
    n_boot=10000,
    random_seed=12345,
):
    """
    Resample independent realizations separately at each network size.

    For each bootstrap sample:
        1. resample realizations at each N
        2. fit A = a/sqrt(N) + b
        3. store a and b
    """

    rng = np.random.default_rng(random_seed)

    n_sizes, n_realizations = A.shape

    slopes = np.empty(n_boot)
    intercepts = np.empty(n_boot)

    for iboot in range(n_boot):

        A_boot = np.empty_like(A)

        for isize in range(n_sizes):

            values = A[isize, :]
            values = values[np.isfinite(values)]

            if len(values) == 0:
                A_boot[isize, :] = np.nan
                continue

            sampled = rng.choice(
                values,
                size=n_realizations,
                replace=True,
            )

            A_boot[isize, :] = sampled

        a, b, _ = fit_free_intercept(N, A_boot)

        slopes[iboot] = a
        intercepts[iboot] = b

    return slopes, intercepts


slopes_085, b_boot_085 = bootstrap_intercept(
    N_085,
    A_085,
    n_boot=N_BOOT,
    random_seed=RANDOM_SEED,
)

slopes_160, b_boot_160 = bootstrap_intercept(
    N_160,
    A_160,
    n_boot=N_BOOT,
    random_seed=RANDOM_SEED + 1,
)
```

```python
# ============================================================
# BOOTSTRAP CONFIDENCE INTERVALS
# ============================================================

def print_bootstrap_summary(label, b_boot):

    low, median, high = np.percentile(
        b_boot,
        [2.5, 50, 97.5]
    )

    print()
    print(label)
    print(f"median b = {median:.6g}")
    print(f"95% bootstrap CI = [{low:.6g}, {high:.6g}]")

    return low, median, high


ci_085 = print_bootstrap_summary(
    "nu_ext / nu_theta = 0.85",
    b_boot_085,
)

ci_160 = print_bootstrap_summary(
    "nu_ext / nu_theta = 1.6",
    b_boot_160,
)
```

```python
# ============================================================
# MAIN FINITE-SIZE FIGURE
# ============================================================

mean_085 = np.nanmean(A_085, axis=1)
std_085 = np.nanstd(A_085, axis=1, ddof=1)

mean_160 = np.nanmean(A_160, axis=1)
std_160 = np.nanstd(A_160, axis=1, ddof=1)

x_085 = 1.0 / np.sqrt(N_085)
x_160 = 1.0 / np.sqrt(N_160)


fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))


# ------------------------------------------------------------
# LEFT: Aosc vs N
# ------------------------------------------------------------

axes[0].errorbar(
    N_085,
    mean_085,
    yerr=std_085,
    fmt="o-",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 0.85$",
)

axes[0].errorbar(
    N_160,
    mean_160,
    yerr=std_160,
    fmt="o-",
    capsize=3,
    label=r"$\nu_{ext}/\nu_\theta = 1.6$",
)

axes[0].set_xlabel("Number of neurons, N")
axes[0].set_ylabel("Oscillation strength (Hz)")
axes[0].grid(True)
axes[0].legend()


# ------------------------------------------------------------
# RIGHT: Aosc vs 1/sqrt(N)
# ------------------------------------------------------------

axes[1].errorbar(
    x_085,
    mean_085,
    yerr=std_085,
    fmt="o",
    capsize=3,
    label=r"$0.85$",
)

axes[1].errorbar(
    x_160,
    mean_160,
    yerr=std_160,
    fmt="o",
    capsize=3,
    label=r"$1.6$",
)


# Fit lines
x_fit = np.linspace(
    min(x_085.min(), x_160.min()),
    max(x_085.max(), x_160.max()),
    200,
)

axes[1].plot(
    x_fit,
    result_085["a"] * x_fit + result_085["b"],
    "-",
    label=rf"$0.85$: $b={result_085['b']:.3f}$",
)

axes[1].plot(
    x_fit,
    result_160["a"] * x_fit + result_160["b"],
    "-",
    label=rf"$1.6$: $b={result_160['b']:.3f}$",
)

axes[1].set_xlabel(r"$1/\sqrt{N}$")
axes[1].set_ylabel("Oscillation strength (Hz)")
axes[1].grid(True)
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# RESIDUALS FROM PURE 1/sqrt(N) SCALING
# ============================================================

pred_085_zero = result_085["a0"] / np.sqrt(N_085)
pred_160_zero = result_160["a0"] / np.sqrt(N_160)

residual_085 = mean_085 - pred_085_zero
residual_160 = mean_160 - pred_160_zero


plt.figure(figsize=(6, 4.5))

plt.axhline(0, linestyle="--", linewidth=1)

plt.plot(
    N_085,
    residual_085,
    "o-",
    label=r"$\nu_{ext}/\nu_\theta = 0.85$",
)

plt.plot(
    N_160,
    residual_160,
    "o-",
    label=r"$\nu_{ext}/\nu_\theta = 1.6$",
)

plt.xlabel("Number of neurons, N")
plt.ylabel(
    r"Residual from $A = a/\sqrt{N}$"
)
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
