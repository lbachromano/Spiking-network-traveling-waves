# Beta attenuation latency maps and planarity

Python translation of the MATLAB code supplied in this conversation. Input files are `Rec_x.mat`, `Rec_y.mat`, and `LFP_absSum_i<N>_J1.mat`, each containing `data`. LFP data must be **time × channels**; coordinates are in **metres**.

Run the cells in order. Dependencies: NumPy, SciPy, pandas, and Matplotlib (`%pip install numpy scipy pandas matplotlib` if needed).

### Translation details
- The first 100 samples are removed. Subsequent sample windows retain the original MATLAB indexing **on the trimmed signal**; `tref_ms` is not shifted to compensate. The original indexing assumes 1000 Hz, which is enforced below.
- Latency is the time of the **first sample below threshold**, following a downward crossing after the detected population peak. No interpolation is added.
- The supplied plane-fitting function uses an **overall regression F-test**, not a permutation test. Its treatment of near-perfect fits (`SSE <= eps` gives `p = NaN`) is retained.
- The supplied `buttfilt` uses Butterworth coefficients and forward–backward filtering. Here SciPy `filtfilt` uses odd padding with length `3*(max(len(a),len(b))-1)`. The separate MATLAB `filtfilt2` source was not supplied, so exact edge handling and numerical equality cannot be verified.
- MATLAB `fillmissing(A, 'movmean', [3 3])` is a one-dimensional window: three positions before and after, along the first nonsingleton dimension. It is **not** a 3 × 3 spatial smoother. That behavior is retained for the optional map display, and does not enter the plane fit.
- The MAT output stores `planarity_results` as a **struct of column arrays**, rather than a MATLAB table. The CSV has the original columns.
- Sixteen polar bins span 360°, so each bin is **22.5°**, despite the original 20° comment. Jitter is reproducible in Python but need not match MATLAB's random sequence.

No experimental data were supplied, so this notebook has not been run on your trials.


```python
from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal, stats
from scipy.io import loadmat, savemat
```

## Parameters
Set `folder_name` to the folder containing the MAT files. The other defaults reproduce the supplied script.

```python
folder_name = Path(Supplementary_Rx150_examples_140")
trial_numbers = np.arange(1, 140)

minimum_valid_fraction = 0.8
minimum_latency_span_ms = 2
minimum_R2 = 0.2
planarity_alpha = 0.05

threshold = 0.8
win_ms = np.arange(-400, 200) #Select where to search for the highest peak of the
                                #mean beta envelope.
order = 4
fs = 1000
skip_initial_window = 100  # samples; 100 ms at fs=1000 Hz
# This value is used as a MATLAB-style sample index AFTER trimming.
tref_ms = 1200
width_filter = 3
beta_range = (12, 30)
start_ms = -10
end_ms = 200

psd_start = tref_ms - 400 - 1
psd_stop = tref_ms + 200
#psd_start = tref_ms -900 - 1
#psd_stop = tref_ms 

	#Select the LFP segment used to estimate the dominant beta frequency,
                        #which sets the filter band.

if fs != 1000:
    raise ValueError("Original time-window indexing assumes fs=1000 Hz.")
```

## Filtering, outliers, missing values, and plane fitting

```python
def load_data(path):
    """Read the variable 'data' from a standard (non-v7.3) MATLAB file."""
    contents = loadmat(path)
    if "data" not in contents:
        raise KeyError(f"{path}: variable 'data' is missing")
    return np.asarray(contents["data"], dtype=float)


def buttfilt(dat, freqrange, samplerate=None, filttype="stop", order=4):
    """Butterworth wrapper; returns (filtered_data, list_of_(b,a)_pairs).

    To reuse coefficients, pass the returned list as freqrange and omit
    samplerate. Filter rows, except that a column vector is filtered on axis 0.
    SciPy filtfilt replaces the unavailable MATLAB filtfilt2 implementation.
    """
    dat = np.asarray(dat, dtype=float)
    if samplerate is None:
        filters = freqrange
    else:
        freqrange = np.asarray(freqrange, dtype=float)
        if filttype == "bandpass" and freqrange.size == 2:
            freqrange = freqrange.ravel()
            if freqrange[0] <= 0:
                filttype, freqrange = "low", np.asarray(freqrange[1])
            elif np.isinf(freqrange[1]):
                filttype, freqrange = "high", np.asarray(freqrange[0])
        btype = {"stop": "bandstop", "low": "lowpass",
                 "high": "highpass", "bandpass": "bandpass"}[filttype]
        filters = []
        for band in np.atleast_2d(freqrange):
            cutoff = float(band[0]) if band.size == 1 else band
            b, a = signal.butter(order, cutoff / (samplerate / 2), btype=btype)
            filters.append((b, a))
    axis = 0 if dat.ndim == 2 and dat.shape[1] == 1 else -1
    for b, a in filters:
        padlen = 3 * (max(len(a), len(b)) - 1)
        dat = signal.filtfilt(b, a, dat, axis=axis, padtype="odd", padlen=padlen)
    return dat, filters


def median_outlier_mask(values):
    """MATLAB isoutlier(values, 'median'): >3 scaled MAD from median."""
    values = np.asarray(values, dtype=float)
    mask = np.zeros(values.shape, dtype=bool)
    finite = np.isfinite(values)
    if finite.any():
        center = np.median(values[finite])
        mad = np.median(np.abs(values[finite] - center))
        scaled_mad = mad / stats.norm.ppf(0.75)
        mask[finite] = np.abs(values[finite] - center) > 3 * scaled_mad
    return mask


def fillmissing_movmean(values, before=3, after=3):
    """Fill NaNs from ORIGINAL neighbors along first nonsingleton dimension."""
    values = np.asarray(values, dtype=float)
    axis = next((i for i, size in enumerate(values.shape) if size > 1), 0)
    source = np.moveaxis(values, axis, 0)
    filled = source.copy()
    for index in range(source.shape[0]):
        window = source[max(0, index-before):index+after+1]
        count = np.sum(~np.isnan(window), axis=0)
        total = np.nansum(window, axis=0)
        average = np.divide(total, count, out=np.full_like(total, np.nan),
                            where=count > 0)
        filled[index] = np.where(np.isnan(source[index]), average, source[index])
    return np.moveaxis(filled, 0, axis)


def fit_latency_plane(x_mm, y_mm, latency_ms, valid_channels,
                      minimum_valid_fraction, minimum_R2,
                      minimum_latency_span_ms, alpha):
    """Fit latency = intercept + gx*x + gy*y; direction is early to late."""
    x_mm = np.asarray(x_mm, dtype=float).ravel()
    y_mm = np.asarray(y_mm, dtype=float).ravel()
    latency_ms = np.asarray(latency_ms, dtype=float).ravel()
    valid = (np.asarray(valid_channels, dtype=bool).ravel()
             & np.isfinite(x_mm) & np.isfinite(y_mm) & np.isfinite(latency_ms))
    number_total_channels = latency_ms.size
    number_valid = np.count_nonzero(valid)
    result = dict(
        is_planar=False, direction_rad=np.nan, direction_deg=np.nan,
        R2=np.nan, p_value=np.nan, gradient_x_ms_per_mm=np.nan,
        gradient_y_ms_per_mm=np.nan, gradient_ms_per_mm=np.nan,
        speed_m_per_s=np.nan, number_valid_channels=number_valid,
        latency_span_ms=np.nan,
        fitted_latency_ms=np.full(number_total_channels, np.nan),
    )
    minimum_number_valid = max(6, int(np.ceil(
        minimum_valid_fraction * number_total_channels)))
    if number_valid < minimum_number_valid:
        return result
    x, y, t = x_mm[valid], y_mm[valid], latency_ms[valid]
    latency_span = np.ptp(t)
    result["latency_span_ms"] = latency_span
    if latency_span < minimum_latency_span_ms:
        return result
    design_matrix = np.column_stack((np.ones(number_valid),
                                     x - x.mean(), y - y.mean()))
    coefficients = np.linalg.lstsq(design_matrix, t, rcond=None)[0]
    fitted_t = design_matrix @ coefficients
    SSE = np.sum((t - fitted_t)**2)
    SST = np.sum((t - t.mean())**2)
    eps = np.finfo(float).eps
    if SST <= eps:
        return result
    observed_R2 = max(0.0, 1 - SSE/SST)
    gx, gy = coefficients[1:]
    gradient_magnitude = np.hypot(gx, gy)
    if gradient_magnitude <= eps:
        return result
    theta = np.arctan2(gy, gx)
    degrees_freedom_error = number_valid - 3
    p_value = np.nan
    if degrees_freedom_error > 0 and SSE > eps:
        F_statistic = ((SST - SSE) / 2) / (SSE / degrees_freedom_error)
        if np.isfinite(F_statistic) and F_statistic >= 0:
            # Equivalent F-test; survival function avoids subtracting from 1.
            p_value = stats.f.sf(F_statistic, 2, degrees_freedom_error)
    result.update(
        R2=observed_R2, p_value=p_value, direction_rad=theta,
        direction_deg=np.degrees(theta) % 360,
        gradient_x_ms_per_mm=gx, gradient_y_ms_per_mm=gy,
        gradient_ms_per_mm=gradient_magnitude,
        speed_m_per_s=1 / gradient_magnitude,
        is_planar=bool(observed_R2 >= minimum_R2
                       and np.isfinite(p_value) and p_value <= alpha
                       and latency_span >= minimum_latency_span_ms),
    )
    result["fitted_latency_ms"][valid] = fitted_t
    return result
```

## Analyze trials
Coordinates are loaded once. `trial_maps` retains each trial's raw, masked, and filled latency maps for inspection. Short recordings raise an informative error rather than silently shortening the analysis windows. If no local envelope peak exists, the trial is marked invalid with a warning (the MATLAB script would fail there).

```python
Rec_x = load_data(folder_name / "Rec_x.mat").ravel(order="F")
Rec_y = load_data(folder_name / "Rec_y.mat").ravel(order="F")
if Rec_x.size != Rec_y.size:
    raise ValueError("Rec_x and Rec_y must have the same number of coordinates.")

# Round to 9 decimal places, with ties away from zero as in MATLAB.
def round_coordinates(values):
    return np.sign(values) * np.floor(np.abs(values) * 1e9 + 0.5) / 1e9

xu, yu = round_coordinates(Rec_x), round_coordinates(Rec_y)
coordinate_valid = np.isfinite(xu) & np.isfinite(yu)
xs, ys = np.unique(xu[coordinate_valid]), np.unique(yu[coordinate_valid])
x_R = np.full(Rec_x.size, -1, dtype=int)
y_R = np.full(Rec_y.size, -1, dtype=int)
x_R[coordinate_valid] = np.searchsorted(xs, xu[coordinate_valid])
y_R[coordinate_valid] = np.searchsorted(ys, yu[coordinate_valid])
if not coordinate_valid.any():
    raise ValueError("No finite recording-site coordinates.")
# Also accommodates rectangular grids; matches square-grid MATLAB output.
map_shape = (len(ys), len(xs))

rows = []
trial_maps = {}
trial_diagnostics = []

for trial_number in trial_numbers:
    name = folder_name / f"LFP_absSum_i{trial_number}_J1.mat"
    data = load_data(name)
    if data.ndim != 2 or data.shape[1] != Rec_x.size:
        raise ValueError(f"{name}: expected time × {Rec_x.size} channels; got {data.shape}")
    lfp_temp = data.T[:, skip_initial_window:]
    n_channel, T = lfp_temp.shape

    # MATLAB sample indices 800:1400 become Python indices 799:1400.

    if psd_start < 0 or psd_stop > T:
        raise ValueError(f"Trial {trial_number}: PSD window exceeds trimmed recording.")
    lfp_f = lfp_temp[:, psd_start:psd_stop]
    duration = lfp_f.shape[1]
    nperseg = min(duration, 512)
    noverlap = nperseg // 2
    nfft = 1 << int(np.ceil(np.log2(max(1024, nperseg))))
    x_mean = signal.detrend(np.nanmean(lfp_f, axis=0), type="linear")
    f_psd, Pm = signal.welch(
        x_mean, fs=fs, window=signal.windows.hamming(nperseg, sym=True),
        nperseg=nperseg, noverlap=noverlap, nfft=nfft,
        detrend=False, scaling="density",
    )
    with np.errstate(divide="ignore"):
        Pm_log = 10 * np.log10(Pm)
    beta_idx = (f_psd >= beta_range[0]) & (f_psd <= beta_range[1])
    f_beta, p_beta = f_psd[beta_idx], Pm_log[beta_idx]
    if f_beta.size == 0:
        raise ValueError("No PSD frequency bins in beta_range.")
    peaks, _ = signal.find_peaks(p_beta)
    k = peaks[np.argmax(p_beta[peaks])] if peaks.size else np.argmax(p_beta)
    domFreq_LFP = f_beta[k]
    freqrange = (max(0.5, domFreq_LFP - width_filter),
                 min(fs/2 - 1e-6, domFreq_LFP + width_filter))
    print(f"LFP-only band centered at {domFreq_LFP:.2f} Hz.")

    filtered_lfp, filters = buttfilt(lfp_temp, freqrange, fs, "bandpass", order)
    envelopes = np.abs(signal.hilbert(filtered_lfp, axis=1))
    full_maximum = np.max(envelopes, axis=1, keepdims=True)
    normalized_full_envelopes = np.divide(
        envelopes, full_maximum, out=np.full_like(envelopes, np.nan),
        where=full_maximum != 0,
    )
    # Use zero-based indices consistently from here onward.
    win_samp = tref_ms + win_ms - 1
    if win_samp[0] < 0 or win_samp[-1] >= T:
        raise ValueError(f"Trial {trial_number}: peak-search window exceeds recording.")
    mean_envelope = np.mean(envelopes, axis=0)
    rel_idx, _ = signal.find_peaks(mean_envelope[win_samp])
    latency_ms = np.full(n_channel, np.nan)
    event_sample = None
    if rel_idx.size == 0:
        warnings.warn(f"Trial {trial_number}: no population envelope peak; invalid trial.")
    else:
        selected_index_pk_envelope = rel_idx[np.argmax(mean_envelope[win_samp[rel_idx]])]
        event_sample = int(win_samp[selected_index_pk_envelope])
        start_offset = int(np.sign(start_ms) * np.floor(abs(start_ms)*fs/1000 + 0.5))
        end_offset = int(np.sign(end_ms) * np.floor(abs(end_ms)*fs/1000 + 0.5))
        win_samples = np.arange(event_sample + start_offset,
                                event_sample + end_offset + 1)
        if win_samples[0] < 0 or win_samples[-1] >= T:
            raise ValueError(f"Trial {trial_number}: latency window exceeds recording.")
        time_axis_ms = (win_samples - event_sample) * (1000/fs)
        zoom_env = envelopes[:, win_samples]
        channel_maximum = np.max(zoom_env, axis=1, keepdims=True)
        channel_maximum[channel_maximum == 0] = np.nan
        norm_env = zoom_env / channel_maximum
        event_index_in_window = np.flatnonzero(win_samples >= event_sample)[0]
        for u, channel_envelope in enumerate(norm_env):
            crossings = np.flatnonzero(
                (channel_envelope[event_index_in_window:-1] >= threshold)
                & (channel_envelope[event_index_in_window+1:] < threshold)
            )
            if crossings.size:
                crossing_index = event_index_in_window + crossings[0] + 1
                latency_ms[u] = time_axis_ms[crossing_index]

    LFP_latency_map = np.full(map_shape, np.nan)
    for u in np.flatnonzero(coordinate_valid):
        LFP_latency_map[y_R[u], x_R[u]] = latency_ms[u]
    outlier_mask = median_outlier_mask(latency_ms)
    valid_channels = (np.isfinite(latency_ms) & ~outlier_mask
                      & np.isfinite(Rec_x) & np.isfinite(Rec_y))
    plane_result = fit_latency_plane(
        1000*Rec_x, 1000*Rec_y, latency_ms, valid_channels,
        minimum_valid_fraction, minimum_R2, minimum_latency_span_ms,
        planarity_alpha,
    )
    rows.append(dict(
        trial_id=int(trial_number), is_planar=plane_result["is_planar"],
        direction_rad=plane_result["direction_rad"],
        direction_deg=plane_result["direction_deg"], plane_R2=plane_result["R2"],
        planarity_p=plane_result["p_value"],
        gradient_ms_per_mm=plane_result["gradient_ms_per_mm"],
        speed_m_per_s=plane_result["speed_m_per_s"],
        number_valid_channels=plane_result["number_valid_channels"],
    ))
    print(f"Trial {trial_number}: planar = {int(plane_result['is_planar'])}, "
          f"R2 = {plane_result['R2']:.3f}, p = {plane_result['p_value']:.4f}, "
          f"direction = {plane_result['direction_deg']:.1f} deg, "
          f"valid channels = {plane_result['number_valid_channels']}")
    LFP_latency_masked = LFP_latency_map.copy()
    for u in np.flatnonzero(outlier_mask & coordinate_valid):
        LFP_latency_masked[y_R[u], x_R[u]] = np.nan
    LFP_filled = fillmissing_movmean(LFP_latency_masked, 3, 3)
    trial_maps[int(trial_number)] = dict(
        raw=LFP_latency_map, masked=LFP_latency_masked, filled=LFP_filled,
        latency_ms=latency_ms.copy(), plane_result=plane_result,
    )
    trial_diagnostics.append(dict(
        trial_id=int(trial_number), dominant_frequency_hz=domFreq_LFP,
        # Add 1 only for a diagnostic directly comparable with MATLAB.
        event_sample_matlab=np.nan if event_sample is None else event_sample + 1,
    ))

planarity_results = pd.DataFrame(rows)
planarity_results
```

## Save trial-by-trial results
Like the MATLAB script, this cell overwrites the two named result files in the data folder. In MATLAB, convert the saved struct to a table with `struct2table(S.planarity_results)` after loading it into `S`.

```python
csv_path = folder_name / "beta_attenuation_planarity_results.csv"
mat_path = folder_name / "beta_attenuation_planarity_results.mat"
planarity_results.to_csv(csv_path, index=False)
savemat(mat_path, {"planarity_results": {
    column: planarity_results[column].to_numpy().reshape(-1, 1)
    for column in planarity_results.columns
}}, do_compression=True)
print(f"Saved {csv_path} and {mat_path}")
```

## Polar plots
Directions point from earlier toward later attenuation: **0° = increasing Rec_x**, **90° = increasing Rec_y**, **180° = decreasing Rec_x**, **270° = decreasing Rec_y**. Only planar trials enter these plots.

```python
accepted_directions = planarity_results.loc[
    planarity_results["is_planar"] & np.isfinite(planarity_results["direction_rad"]),
    "direction_rad",
].to_numpy()

fig, ax = plt.subplots(subplot_kw={"projection": "polar"}, figsize=(6, 6))
counts, edges = np.histogram(accepted_directions % (2*np.pi),
                            bins=np.linspace(0, 2*np.pi, 17))
probability = counts / accepted_directions.size if accepted_directions.size else counts.astype(float)
ax.bar(edges[:-1], probability, width=np.diff(edges), align="edge", edgecolor="white")
ax.set_theta_zero_location("E")
ax.set_theta_direction(1)
ax.tick_params(labelsize=16)
plt.show()
print(f"Planar trials: {accepted_directions.size} / {len(planarity_results)}")
```

```python
rng = np.random.default_rng(1)
jitter_amount = 0.1
r_jittered = 1 + (rng.random(accepted_directions.size) - 0.5) * jitter_amount
fig, ax = plt.subplots(subplot_kw={"projection": "polar"}, figsize=(6, 6))
ax.plot(accepted_directions, r_jittered, "o", markerfacecolor=(0.2, 0.6, 0.8),
        markeredgecolor="k", markersize=6)
ax.set_theta_zero_location("E")
ax.set_theta_direction(1)
ax.tick_params(labelsize=16)
ax.set_ylim(0, 1 + jitter_amount)
ax.set_yticks([1])
ax.set_yticklabels([])
plt.show()
```

## Optional spatial maps
Select one analyzed trial. The filled map is for display only; classification uses measured latencies after outlier removal.

```python
trial_to_plot = int(trial_numbers[0])
maps = trial_maps[trial_to_plot]
fig, axes = plt.subplots(1, 3, figsize=(12, 4), constrained_layout=True)
finite_values = maps["raw"][np.isfinite(maps["raw"])]
vmin, vmax = (finite_values.min(), finite_values.max()) if finite_values.size else (0, 1)
if vmin == vmax:
    vmax = vmin + 1
for ax, key, title in zip(axes, ["raw", "masked", "filled"],
                          ["Raw latencies", "Outliers masked", "Missing values filled"]):
    im = ax.imshow(np.ma.masked_invalid(maps[key]), origin="lower",
                   aspect="equal", vmin=vmin, vmax=vmax)
    ax.set_title(title)
    ax.set_xticks([])
    ax.set_yticks([])
fig.colorbar(im, ax=axes, label="Latency from population peak (ms)")
fig.suptitle(f"Trial {trial_to_plot}")
plt.show()
```
