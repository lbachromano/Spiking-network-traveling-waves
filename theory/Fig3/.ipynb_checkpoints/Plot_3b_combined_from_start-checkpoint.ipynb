```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat

 
```

```python
# Load simulated oscillation frequencies
lf = loadmat("Oscillation_combined_rates_I.mat", squeeze_me=True)

all_y = np.asarray(lf["mean_freq"]).squeeze()
all_std = np.asarray(lf["sem_freq"]).squeeze()
all_x = np.asarray(lf["input_drive"]).squeeze()
```

```python
# Load stability results
lc = loadmat("stability_results/26caStability_g_fwdtest_3.625.mat", squeeze_me=True)


list_varying_param = np.asarray(lc["list_varying_param"]).squeeze()
local_minima_all = np.atleast_1d(lc["local_minima_all"])

all_param_list = []
all_minima_list = []

for i, min_h in enumerate(local_minima_all):
    # MATLAB cell contents can arrive as object arrays; convert to numeric matrix
    min_h = np.asarray(min_h, dtype=float)
    min_h = np.atleast_2d(min_h)

    nr_minima = min_h.shape[0]
    params_h = np.full(nr_minima, list_varying_param[i])

    all_param_list.append(params_h)
    all_minima_list.append(min_h)

all_param = np.concatenate(all_param_list)
all_minima = np.vstack(all_minima_list)
```

```python
# Two-panel figure
fig, axes = plt.subplots(2, 1, figsize=(7, 8), sharex=True)

# Top: real part of the eigenvalue
ax = axes[0]
ax.plot(all_param, all_minima[:, 0], "k", linewidth=2)
ax.axhline(0, color="k")
ax.set_ylabel(r"$\mathrm{Re}\,\lambda$", fontsize=14)
ax.set_xlim(0.7, 1.4)
ax.tick_params(labelsize=14)

# Bottom: predicted frequency and simulation results
ax = axes[1]
ax.errorbar(all_x, all_y, yerr=all_std, fmt="o", label="Simulation")
ax.plot(all_param, np.abs(all_minima[:, 1]), "k", linewidth=2,
        label="Stability prediction")

ax.axvline(0.752, color="k")
ax.axvline(1.126, color="k")

ax.set_ylabel(r"$\omega$", fontsize=18)
ax.set_xlabel(r"$\nu_{\mathrm{ext}}/\nu_{\theta}$", fontsize=16)
ax.set_xlim(0.7, 1.4)
ax.set_ylim(5, 40)
ax.tick_params(labelsize=16)
ax.legend()

fig.tight_layout()
plt.show()
```
