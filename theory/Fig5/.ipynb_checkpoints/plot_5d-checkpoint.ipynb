# Recovered external-rate sweeps

`1.npz` through `4.npz` correspond to the four pasted blocks, in order. Each contains 21 external rates (1650–2550 Hz, step 45 Hz), the printed maximum real parts, completed flags, and the printed root counts (100/100).

Only terminal precision is recoverable. Parameters (`g`, `rho_ee`, `velocity`), maximizing coordinates, frequencies, and spatial grids were not printed and are deliberately omitted. Fill in the legend labels below once the block-to-parameter mapping is known.

The supplied plotting function is extended with automatic parameter labels for original sweep files, a single shared zero line, and `plot_sweeps` for overlays. The original completion and missing-root checks are retained.


```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

def plot_max_real_vs_nu_ext(filename, ax=None, label=None, color=None):
    """Plot one sweep; infer a label from saved parameters when available."""
    filename = Path(filename)
    with np.load(filename, allow_pickle=False) as saved:
        data = {key: saved[key].copy() for key in saved.files}

    order = np.argsort(data["nu_ext"])
    nu_ext = data["nu_ext"][order]
    y = data["max_real"][order].copy()
    completed = data["completed"][order].astype(bool)
    valid = completed & np.isfinite(y)
    y[~valid] = np.nan
    incomplete = valid & (data["n_success"][order] < data["n_total"][order])

    if label is None:
        parts = []
        for key, name in [("g", "g"), ("rho_ee", r"$\rho_{EE}$"),
                          ("velocity", "velocity")]:
            if key in data and np.isfinite(data[key].item()):
                parts.append(f"{name}={data[key].item():g}")
        label = ", ".join(parts) if parts else f"Run {filename.stem}"

    if ax is None:
        _, ax = plt.subplots(figsize=(7, 4.5))
    ax.plot(
        nu_ext, y, "o-",
        label=label,
        color=color,
        markersize=4,
        linewidth=2,
    )
    if np.any(incomplete):
        warning = "Some spatial points have no accepted root"
        existing = ax.get_legend_handles_labels()[1]
        ax.scatter(nu_ext[incomplete], y[incomplete], marker="x", s=90,
                   color="red", zorder=4,
                   label=warning if warning not in existing else "_nolegend_")

    # Repeated calls on the same axes share one zero reference line.
    if not any(line.get_gid() == "stability_zero" for line in ax.lines):
        zero = ax.axhline(0, color="black", linestyle="--", linewidth=1)
        zero.set_gid("stability_zero")
    ax.set_xlabel(r"$\nu_{\mathrm{ext}}\;[\mathrm{Hz}]$")
    ax.set_ylabel(r"$\max_{u,v}\mathrm{Re}\,\lambda_{\mathrm{dom}}\;[\mathrm{s}^{-1}]$")
    ax.legend()
    ax.figure.tight_layout()
    return ax.figure, ax, data


def plot_sweeps(filenames, labels=None, colors=None, ax=None):
    """Overlay multiple saved sweeps with optional labels and colors."""
    filenames = list(filenames)

    if not filenames:
        raise ValueError("No result files selected.")

    labels = {} if labels is None else labels

    if colors is not None and len(colors) != len(filenames):
        raise ValueError("Number of colors must match number of files.")

    if ax is None:
        fig, ax = plt.subplots(figsize=(7, 4.5))
    else:
        fig = ax.figure
    all_data = {}

    for i, filename in enumerate(filenames):
        filename = Path(filename)

        color = None if colors is None else colors[i]

        _, _, data = plot_max_real_vs_nu_ext(
            filename,
            ax=ax,
            label=labels.get(filename.name),
            color=color,
        )

        all_data[str(filename)] = data

    return fig, ax, all_data
```

```python
Results_5d_DIR = Path("Results_5d")

rho = 0.6
g_values = [3.0, 3.3, 3.5, 3.6, 3.9]
files = [
    Results_5d_DIR / f"max_real_vs_nu_ext_g{g:.1f}_rho{rho:.1f}.npz"
    for g in g_values
]
 

labels = {
    path.name: rf"$g = {g:.1f}$"
    for path, g in zip(files, g_values)
}

colors = np.linspace(
    [0.68, 0.85, 0.95],  # Light blue
    [0.02, 0.08, 0.35],  # Dark navy blue
    len(files)
)

fig, ax, Results_5d = plot_sweeps(files, labels=labels, colors=colors)


ax.set_box_aspect(1)
ax.set_xlim(1650,2650)
ax.set_ylim(-27,40)
ax.tick_params(axis="both", labelsize=14)  # numbers on both axes
fig.tight_layout()
# fig.savefig("Max_lambda.svg", format="svg", bbox_inches="tight")
plt.show()
```

```python
Results_5d_DIR = Path("Results_5d")

rho_values = [0.0, 0.6, 0.7, 0.9]
g = 3.6
files = [
    Results_5d_DIR / f"max_real_vs_nu_ext_g{g:.1f}_rho{rho:.1f}.npz"
    for rho in rho_values
]
 

labels = {
    path.name: rf"$\rho = {rho:.1f}$"
    for path, rho in zip(files, rho_values)
}

colors = np.linspace(
    [0.85, 0.72, 0.70],  # Light wine  
    [0.42, 0.10, 0.16],  # burgundy
    len(files)
)

fig, ax, Results_5d = plot_sweeps(files, labels=labels, colors=colors)


ax.set_box_aspect(1)
ax.set_xlim(1650,2650)
#ax.set_ylim(-27,40)
ax.tick_params(axis="both", labelsize=14)  # numbers on both axes
fig.tight_layout()
# fig.savefig("Max_lambda.svg", format="svg", bbox_inches="tight")
plt.show()
```

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

Results_5d_DIR = Path("Results_5d")

# One square figure, two vertically stacked subplots
fig, axes = plt.subplots(
    2, 1,
    figsize=(7, 7),
    sharex=True,
    gridspec_kw={"hspace": 0}
)

# ==================================================
# TOP: varying g, fixed rho = 0.6
# ==================================================

rho = 0.6
g_values = [3.0, 3.3, 3.5, 3.6, 3.9]

files_g = [
    Results_5d_DIR / f"max_real_vs_nu_ext_g{g:.1f}_rho{rho:.1f}.npz"
    for g in g_values
]

labels_g = {
    path.name: rf"$g = {g:.1f}$"
    for path, g in zip(files_g, g_values)
}

colors_g = np.linspace(
    [0.68, 0.85, 0.95],
    [0.02, 0.08, 0.35],
    len(files_g)
)

plot_sweeps(
    files_g, ax=axes[0],
    labels=labels_g, colors=colors_g
)

axes[0].set_ylim(-27, 40)

# ==================================================
# BOTTOM: varying rho, fixed g = 3.6
# ==================================================

rho_values = [0.0, 0.6, 0.7, 0.9]
g = 3.6

files_rho = [
    Results_5d_DIR / f"max_real_vs_nu_ext_g{g:.1f}_rho{rho:.1f}.npz"
    for rho in rho_values
]

labels_rho = {
    path.name: rf"$\rho = {rho:.1f}$"
    for path, rho in zip(files_rho, rho_values)
}

colors_rho = np.linspace(
    [0.90, 0.72, 0.75],
    [0.42, 0.10, 0.16],
    len(files_rho)
)

plot_sweeps(
    files_rho, ax=axes[1],
    labels=labels_rho, colors=colors_rho
)

# ==================================================
# SHARED AXES AND FORMATTING
# ==================================================

for ax in axes:
    ax.set_xlim(1650, 2650)
    ax.tick_params(axis="both", labelsize=14)

# Remove x-axis label and tick labels from top panel
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", labelbottom=False)

# Only bottom panel has x-axis label
axes[1].xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{x / 2195:.3g}")
)
axes[1].set_xlabel(r"$\nu_{\mathrm{ext}} / \nu_{\mathrm{\theta}}$", fontsize=16)

# Remove individual y-axis labels
axes[0].set_ylabel("")
axes[1].set_ylabel("")

# One shared y-axis label
fig.supylabel(
    r"$\max_{u,v}\operatorname{Re}\lambda$",
    fontsize=16,
    x=0.02
)

# Square overall figure
fig.subplots_adjust(
    left=0.17,
    right=0.95,
    bottom=0.12,
    top=0.97,
    hspace=0
)

fig.savefig(
    "Max_lambda_combined.svg",
    format="svg",
    bbox_inches="tight"
)

plt.show()
```
